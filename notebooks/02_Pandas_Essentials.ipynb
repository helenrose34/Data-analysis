# Pandas Essentials for Data Analysis

## Introduction to Pandas
Pandas is the primary tool for data manipulation and analysis in Python.

```python
# Import pandas
import pandas as pd
import numpy as np

print('Pandas version:', pd.__version__)
```

## 1. Series - One-Dimensional Data

A Series is a one-dimensional array with labels.

```python
# Creating a Series
temperatures = pd.Series([20, 22, 19, 25, 23], name='Temperature (°C)')
print(temperatures)
print()

# Accessing elements
print('First temperature:', temperatures[0])
print('Mean temperature:', temperatures.mean())
```

## 2. DataFrames - Two-Dimensional Data

A DataFrame is like a table with rows and columns.

```python
# Creating a DataFrame
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana'],
    'Age': [25, 30, 35, 28],
    'Salary': [50000, 60000, 75000, 55000],
    'Department': ['IT', 'HR', 'Finance', 'IT']
}

df = pd.DataFrame(data)
print(df)
print()
print('DataFrame Info:')
print(df.info())
```

## 3. Basic Data Exploration

Explore your data quickly.

```python
# View first few rows
print('First 2 rows:')
print(df.head(2))
print()

# Statistics
print('Basic Statistics:')
print(df.describe())
print()

# Column names and shape
print('Columns:', df.columns.tolist())
print('Shape (rows, columns):', df.shape)
```

## 4. Selecting Data

Different ways to access data.

```python
# Select a single column
print('Names column:')
print(df['Name'])
print()

# Select multiple columns
print('Name and Age:')
print(df[['Name', 'Age']])
print()

# Select rows by position
print('First row:')
print(df.iloc[0])
print()

# Filter by condition
print('People older than 28:')
print(df[df['Age'] > 28])
```

## 5. Data Cleaning

Handle missing data and duplicates.

```python
# Create a DataFrame with missing values
data_with_missing = {
    'Name': ['Alice', 'Bob', None, 'Diana'],
    'Age': [25, None, 35, 28],
    'Score': [90, 85, 88, 92]
}
df_missing = pd.DataFrame(data_with_missing)
print('DataFrame with missing values:')
print(df_missing)
print()

# Check for missing values
print('Missing values:')
print(df_missing.isnull())
print()

# Fill missing values
df_filled = df_missing.fillna({'Age': df_missing['Age'].mean()})
print('After filling missing Age with mean:')
print(df_filled)
```

## 6. Data Transformation

Modify and create new columns.

```python
# Create new columns
df['Salary_in_thousands'] = df['Salary'] / 1000

# Apply functions to columns
df['Name_Length'] = df['Name'].apply(len)

# Use conditions to create new columns
df['Salary_Range'] = df['Salary'].apply(lambda x: 'Low' if x < 60000 else 'High')

print(df)
```

## 7. Grouping and Aggregation

Summarize data by groups.

```python
# Group by Department and calculate mean salary
print('Average salary by department:')
print(df.groupby('Department')['Salary'].mean())
print()

# Multiple aggregations
print('Salary statistics by department:')
print(df.groupby('Department')['Salary'].agg(['mean', 'min', 'max', 'count']))
```
