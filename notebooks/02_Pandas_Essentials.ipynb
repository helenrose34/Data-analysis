# Pandas Essentials for Data Analysis

## Introduction to Pandas

Pandas is the primary tool for data manipulation and analysis in Python. It provides:
- **DataFrames**: 2D labeled data structures
- **Series**: 1D labeled data
- Powerful data manipulation and cleaning tools

## Setup

```python
import pandas as pd
import numpy as np

# Display options
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)
```

## 1. Creating DataFrames

```python
# Create from dictionary
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Age': [25, 30, 35, 28],
    'Salary': [50000, 60000, 75000, 55000],
    'Department': ['Sales', 'IT', 'Finance', 'Sales']
}

df = pd.DataFrame(data)
print("DataFrame:")
print(df)
```

```python
# Create from list of lists
data_list = [
    ['Alice', 25, 50000],
    ['Bob', 30, 60000],
    ['Charlie', 35, 75000]
]

df2 = pd.DataFrame(data_list, columns=['Name', 'Age', 'Salary'])
print(df2)
```

## 2. Exploring DataFrames

```python
# Basic information
print("Shape:", df.shape)
print("\nColumn names:", df.columns.tolist())
print("\nData types:\n", df.dtypes)
print("\nFirst 2 rows:")
print(df.head(2))
print("\nLast 2 rows:")
print(df.tail(2))
```

```python
# Statistical summary
print(df.describe())
```

```python
# Info about DataFrame
print(df.info())
```

## 3. Selecting Data

```python
# Select single column (returns Series)
print("Names:")
print(df['Name'])
print("Type:", type(df['Name']))
```

```python
# Select multiple columns
print("Name and Salary:")
print(df[['Name', 'Salary']])
```

```python
# Select by row index
print("First row (index 0):")
print(df.loc[0])
print("\nRows 1 and 2:")
print(df.iloc[1:3])
```

```python
# Conditional selection
print("Employees with salary > 55000:")
print(df[df['Salary'] > 55000])
print("\nEmployees in Sales department:")
print(df[df['Department'] == 'Sales'])
```

## 4. Data Manipulation

```python
# Add new column
df['Bonus'] = df['Salary'] * 0.1
print(df)
```

```python
# Remove columns
df_temp = df.drop('Bonus', axis=1)
print(df_temp)
```

```python
# Rename columns
df_renamed = df.rename(columns={'Name': 'Employee', 'Salary': 'Annual_Salary'})
print(df_renamed)
```

```python
# Fill missing values
df_missing = pd.DataFrame({
    'A': [1, 2, np.nan, 4],
    'B': [5, np.nan, np.nan, 8]
})
print("Original:")
print(df_missing)
print("\nFilled with 0:")
print(df_missing.fillna(0))
print("\nForward fill:")
print(df_missing.fillna(method='ffill'))
```

## 5. Grouping and Aggregation

```python
# Group by department
print("Average salary by department:")
print(df.groupby('Department')['Salary'].mean())

print("\nCount by department:")
print(df.groupby('Department').size())
```

```python
# Multiple aggregations
print("Salary statistics by department:")
print(df.groupby('Department')['Salary'].agg(['mean', 'min', 'max', 'count']))
```

## 6. Sorting and Filtering

```python
# Sort by salary
print("Sorted by salary (descending):")
print(df.sort_values('Salary', ascending=False))

print("\nSorted by multiple columns:")
print(df.sort_values(['Department', 'Salary'], ascending=[True, False]))
```

## 7. Working with CSV Files

```python
# Write to CSV
df.to_csv('employees.csv', index=False)
print("Data written to employees.csv")

# Read from CSV
df_loaded = pd.read_csv('employees.csv')
print("\nLoaded data:")
print(df_loaded)
```
