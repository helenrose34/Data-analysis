# Data Visualization

## Transform Data into Insights with Visualizations

Beautiful visualizations help you:
- Discover patterns and trends
- Communicate insights effectively
- Explore data interactively

## Setup

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set style
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 6)
%matplotlib inline
```

## 1. Basic Line Plot

```python
# Sample data
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
sales = [20000, 23000, 19000, 25000, 28000, 31000]

plt.figure(figsize=(10, 6))
plt.plot(months, sales, marker='o', linewidth=2, markersize=8, color='#667eea')
plt.title('Monthly Sales Trend', fontsize=14, fontweight='bold')
plt.xlabel('Month', fontsize=12)
plt.ylabel('Sales ($)', fontsize=12)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## 2. Bar Charts

```python
# Create sample data
departments = ['Sales', 'IT', 'Finance', 'HR']
employees = [25, 30, 15, 10]

plt.figure(figsize=(10, 6))
bars = plt.bar(departments, employees, color=['#667eea', '#764ba2', '#f093fb', '#4facfe'])
plt.title('Employees by Department', fontsize=14, fontweight='bold')
plt.xlabel('Department', fontsize=12)
plt.ylabel('Number of Employees', fontsize=12)

# Add value labels on bars
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2., height,
            f'{int(height)}',
            ha='center', va='bottom')
plt.tight_layout()
plt.show()
```

## 3. Scatter Plots

```python
# Correlation between age and salary
age = np.array([25, 30, 35, 28, 32, 40, 29, 38, 45, 33])
salary = np.array([50000, 55000, 65000, 60000, 62000, 75000, 58000, 70000, 80000, 63000])

plt.figure(figsize=(10, 6))
plt.scatter(age, salary, s=100, alpha=0.6, color='#667eea')
plt.title('Age vs Salary', fontsize=14, fontweight='bold')
plt.xlabel('Age', fontsize=12)
plt.ylabel('Salary ($)', fontsize=12)

# Add trend line
z = np.polyfit(age, salary, 1)
p = np.poly1d(z)
plt.plot(age, p(age), "r--", linewidth=2, label='Trend')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## 4. Histograms

```python
# Distribution of salaries
salaries = np.random.normal(60000, 15000, 200)

plt.figure(figsize=(10, 6))
plt.hist(salaries, bins=20, color='#667eea', alpha=0.7, edgecolor='black')
plt.title('Salary Distribution', fontsize=14, fontweight='bold')
plt.xlabel('Salary ($)', fontsize=12)
plt.ylabel('Frequency', fontsize=12)
plt.grid(True, alpha=0.3, axis='y')
plt.tight_layout()
plt.show()
```

## 5. Box Plots

```python
# Create sample data
dept_data = {
    'Sales': np.random.normal(55000, 8000, 30),
    'IT': np.random.normal(70000, 10000, 30),
    'Finance': np.random.normal(65000, 9000, 30),
    'HR': np.random.normal(50000, 7000, 30)
}

plt.figure(figsize=(10, 6))
plt.boxplot([dept_data['Sales'], dept_data['IT'], dept_data['Finance'], dept_data['HR']],
           labels=['Sales', 'IT', 'Finance', 'HR'])
plt.title('Salary Distribution by Department', fontsize=14, fontweight='bold')
plt.ylabel('Salary ($)', fontsize=12)
plt.grid(True, alpha=0.3, axis='y')
plt.tight_layout()
plt.show()
```

## 6. Pie Charts

```python
# Market share
companies = ['Company A', 'Company B', 'Company C', 'Company D']
market_share = [30, 25, 25, 20]
colors = ['#667eea', '#764ba2', '#f093fb', '#4facfe']

plt.figure(figsize=(10, 8))
plt.pie(market_share, labels=companies, autopct='%1.1f%%', colors=colors,
       startangle=90, textprops={'fontsize': 11})
plt.title('Market Share', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

## 7. Heatmaps (Seaborn)

```python
# Correlation matrix
data = pd.DataFrame({
    'Age': np.random.randint(25, 50, 100),
    'Salary': np.random.randint(50000, 100000, 100),
    'Experience': np.random.randint(1, 25, 100),
    'Performance': np.random.randint(60, 100, 100)
})

correlation = data.corr()

plt.figure(figsize=(8, 6))
sns.heatmap(correlation, annot=True, cmap='coolwarm', center=0,
           fmt='.2f', square=True, cbar_kws={'label': 'Correlation'})
plt.title('Correlation Heatmap', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

## 8. Subplots (Multiple plots)

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Plot 1: Line plot
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
sales = [20000, 23000, 19000, 25000, 28000, 31000]
axes[0, 0].plot(months, sales, marker='o', color='#667eea', linewidth=2)
axes[0, 0].set_title('Sales Trend')
axes[0, 0].set_ylabel('Sales ($)')

# Plot 2: Bar chart
departments = ['Sales', 'IT', 'Finance', 'HR']
employees = [25, 30, 15, 10]
axes[0, 1].bar(departments, employees, color='#764ba2')
axes[0, 1].set_title('Employees by Department')
axes[0, 1].set_ylabel('Count')

# Plot 3: Histogram
ages = np.random.normal(35, 10, 200)
axes[1, 0].hist(ages, bins=15, color='#f093fb', alpha=0.7)
axes[1, 0].set_title('Age Distribution')
axes[1, 0].set_xlabel('Age')
axes[1, 0].set_ylabel('Frequency')

# Plot 4: Scatter plot
age = np.array([25, 30, 35, 28, 32, 40, 29, 38, 45, 33])
salary = np.array([50000, 55000, 65000, 60000, 62000, 75000, 58000, 70000, 80000, 63000])
axes[1, 1].scatter(age, salary, color='#4facfe', s=100, alpha=0.6)
axes[1, 1].set_title('Age vs Salary')
axes[1, 1].set_xlabel('Age')
axes[1, 1].set_ylabel('Salary ($)')

plt.tight_layout()
plt.show()
```
