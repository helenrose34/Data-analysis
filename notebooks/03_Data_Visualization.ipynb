# Data Visualization

## Transform Data into Insights with Visualizations
Learn to create beautiful and meaningful plots.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Set style for better-looking plots
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (10, 6)
```

## 1. Line Charts

Perfect for showing trends over time.

```python
# Sample data
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
sales = [50, 60, 55, 75, 85, 95]

plt.figure(figsize=(10, 6))
plt.plot(months, sales, marker='o', linewidth=2, markersize=8, color='#667eea')
plt.title('Monthly Sales Trend', fontsize=14, fontweight='bold')
plt.xlabel('Month')
plt.ylabel('Sales ($1000s)')
plt.grid(True, alpha=0.3)
plt.show()
```

## 2. Bar Charts

Compare values across categories.

```python
# Sample data
categories = ['Product A', 'Product B', 'Product C', 'Product D']
values = [45, 38, 52, 41]

plt.figure(figsize=(10, 6))
bars = plt.bar(categories, values, color=['#667eea', '#764ba2', '#f093fb', '#4facfe'])
plt.title('Sales by Product', fontsize=14, fontweight='bold')
plt.ylabel('Sales ($1000s)')

# Add value labels on bars
for bar, value in zip(bars, values):
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 1, str(value), ha='center')
plt.show()
```

## 3. Scatter Plots

Show relationships between two variables.

```python
# Sample data
np.random.seed(42)
x = np.random.randn(100)
y = 2 * x + np.random.randn(100)

plt.figure(figsize=(10, 6))
plt.scatter(x, y, alpha=0.6, s=100, color='#667eea')
plt.title('Relationship between X and Y', fontsize=14, fontweight='bold')
plt.xlabel('X Variable')
plt.ylabel('Y Variable')
plt.grid(True, alpha=0.3)
plt.show()
```

## 4. Histograms

Show distribution of data.

```python
# Sample data
data = np.random.normal(loc=100, scale=15, size=1000)

plt.figure(figsize=(10, 6))
plt.hist(data, bins=30, color='#667eea', alpha=0.7, edgecolor='black')
plt.title('Distribution of Test Scores', fontsize=14, fontweight='bold')
plt.xlabel('Score')
plt.ylabel('Frequency')
plt.axvline(data.mean(), color='red', linestyle='--', linewidth=2, label=f'Mean: {data.mean():.1f}')
plt.legend()
plt.show()
```

## 5. Box Plots

Compare distributions across groups.

```python
# Sample data
group_a = np.random.normal(loc=100, scale=15, size=100)
group_b = np.random.normal(loc=110, scale=12, size=100)
group_c = np.random.normal(loc=95, scale=18, size=100)

plt.figure(figsize=(10, 6))
plt.boxplot([group_a, group_b, group_c], labels=['Group A', 'Group B', 'Group C'])
plt.title('Score Distribution by Group', fontsize=14, fontweight='bold')
plt.ylabel('Score')
plt.grid(True, alpha=0.3)
plt.show()
```

## 6. Heatmaps

Visualize correlation matrices.

```python
# Sample data
df = pd.DataFrame({
    'Age': np.random.randint(20, 60, 20),
    'Salary': np.random.randint(30000, 100000, 20),
    'Experience': np.random.randint(0, 30, 20),
    'Performance': np.random.randint(50, 100, 20)
})

correlation = df.corr()

plt.figure(figsize=(8, 6))
sns.heatmap(correlation, annot=True, cmap='coolwarm', center=0, fmt='.2f')
plt.title('Correlation Matrix', fontsize=14, fontweight='bold')
plt.show()
```

## 7. Multiple Subplots

Create multiple plots in one figure.

```python
fig, axes = plt.subplots(2, 2, figsize=(12, 10))

# Plot 1: Line chart
axes[0, 0].plot([1, 2, 3, 4], [10, 20, 15, 25], marker='o', color='#667eea')
axes[0, 0].set_title('Line Chart')

# Plot 2: Bar chart
axes[0, 1].bar(['A', 'B', 'C'], [5, 10, 8], color='#764ba2')
axes[0, 1].set_title('Bar Chart')

# Plot 3: Scatter plot
axes[1, 0].scatter([1, 2, 3], [1, 4, 9], color='#f093fb', s=100)
axes[1, 0].set_title('Scatter Plot')

# Plot 4: Histogram
axes[1, 1].hist([1, 2, 2, 3, 3, 3, 4], bins=4, color='#4facfe')
axes[1, 1].set_title('Histogram')

plt.tight_layout()
plt.show()
```
