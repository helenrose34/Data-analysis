# Python Basics for Data Analysis

## Welcome! 👋
This notebook covers fundamental Python concepts you need for data analysis.

## 1. Variables and Data Types

### What are variables?
Variables are containers that store data values.

```python
# Creating variables
name = 'Alice'
age = 25
height = 5.6
is_student = True

print(f'Name: {name}')
print(f'Age: {age}')
print(f'Height: {height}')
print(f'Is Student: {is_student}')
```

## 2. Data Types

Python has several built-in data types:

```python
# Different data types
integer = 42
floating = 3.14
string = 'Data Analysis'
boolean = True

print(f'Integer: {type(integer)}')
print(f'Float: {type(floating)}')
print(f'String: {type(string)}')
print(f'Boolean: {type(boolean)}')
```

## 3. Lists - Ordered Collections

Lists can hold multiple items in a single variable.

```python
# Creating and working with lists
numbers = [1, 2, 3, 4, 5]
fruits = ['apple', 'banana', 'orange']

print(f'Numbers: {numbers}')
print(f'First number: {numbers[0]}')
print(f'Last number: {numbers[-1]}')

# Adding items
fruits.append('grape')
print(f'Updated fruits: {fruits}')
```

## 4. Dictionaries - Key-Value Pairs

Dictionaries store data with labels (keys).

```python
# Creating and working with dictionaries
person = {
    'name': 'John',
    'age': 30,
    'city': 'New York'
}

print(f'Person: {person}')
print(f'Name: {person["name"]}')
print(f'Age: {person["age"]}')

# Adding new key
person['job'] = 'Engineer'
print(f'Updated person: {person}')
```

## 5. Functions

Functions are reusable blocks of code.

```python
# Defining a function
def greet(name):
    return f'Hello, {name}! Welcome to Data Analysis!'

# Using the function
message = greet('Alice')
print(message)

# Function with multiple parameters
def calculate_average(a, b, c):
    return (a + b + c) / 3

avg = calculate_average(10, 20, 30)
print(f'Average: {avg}')
```

## 6. Loops

Loops allow you to repeat code.

```python
# For loop
fruits = ['apple', 'banana', 'orange']
for fruit in fruits:
    print(f'I like {fruit}')

print('---')

# While loop
count = 0
while count < 5:
    print(f'Count: {count}')
    count += 1
```

## 7. Conditional Statements

Make decisions in your code.

```python
# If-else statements
age = 25

if age >= 18:
    print('You are an adult')
else:
    print('You are a minor')

# Multiple conditions
score = 85
if score >= 90:
    grade = 'A'
elif score >= 80:
    grade = 'B'
elif score >= 70:
    grade = 'C'
else:
    grade = 'F'

print(f'Score: {score}, Grade: {grade}')
```
