## 1. Variables and Data Types

Python has several fundamental data types:

```python
# Integers
age = 25
print(f"Age: {age}, Type: {type(age)}")

# Floats
height = 5.9
print(f"Height: {height}, Type: {type(height)}")

# Strings
name = "Alice"
print(f"Name: {name}, Type: {type(name)}")

# Booleans
is_student = True
print(f"Is Student: {is_student}, Type: {type(is_student)}")
```

## 2. Collections

### Lists (ordered, mutable)

```python
# Creating and manipulating lists
numbers = [1, 2, 3, 4, 5]
print(f"List: {numbers}")
print(f"First element: {numbers[0]}")
print(f"Last element: {numbers[-1]}")

# List operations
numbers.append(6)
print(f"After append: {numbers}")

# List slicing
print(f"Slice [1:4]: {numbers[1:4]}")
```

### Tuples (ordered, immutable)

```python
# Tuples are immutable (cannot be changed)
coordinates = (10, 20)
print(f"Coordinates: {coordinates}")
x, y = coordinates
print(f"X: {x}, Y: {y}")
```

### Dictionaries (key-value pairs)

```python
# Dictionaries store data with keys and values
person = {
    "name": "Bob",
    "age": 30,
    "city": "New York"
}

print(f"Name: {person['name']}")
print(f"Age: {person['age']}")

# Adding new key-value
person['job'] = "Data Analyst"
print(person)
```

## 3. Control Flow

### If/Else Statements

```python
score = 85

if score >= 90:
    grade = 'A'
elif score >= 80:
    grade = 'B'
elif score >= 70:
    grade = 'C'
else:
    grade = 'F'

print(f"Score: {score}, Grade: {grade}")
```

### Loops

```python
# For loop
print("For Loop:")
for i in range(5):
    print(f"  i = {i}")

# While loop
print("\nWhile Loop:")
count = 0
while count < 3:
    print(f"  count = {count}")
    count += 1

# Loop through list
print("\nIterating over list:")
fruits = ['apple', 'banana', 'cherry']
for fruit in fruits:
    print(f"  {fruit}")
```

## 4. Functions

```python
def greet(name, greeting="Hello"):
    """
    Greet a person with optional custom greeting.
    """
    return f"{greeting}, {name}!"

# Function calls
print(greet("Alice"))
print(greet("Bob", greeting="Hi"))

# Function with multiple returns
def calculate_stats(numbers):
    return min(numbers), max(numbers), sum(numbers) / len(numbers)

data = [10, 20, 30, 40, 50]
minimum, maximum, average = calculate_stats(data)
print(f"Min: {minimum}, Max: {maximum}, Avg: {average}")
```

## 5. List Comprehensions

```python
# Traditional way
squares_traditional = []
for i in range(5):
    squares_traditional.append(i**2)
print(f"Traditional: {squares_traditional}")

# List comprehension (more Pythonic)
squares_compact = [i**2 for i in range(5)]
print(f"List comprehension: {squares_compact}")

# Conditional list comprehension
even_squares = [i**2 for i in range(10) if i % 2 == 0]
print(f"Even squares: {even_squares}")
```

## 6. File Operations

```python
# Writing to a file
with open('sample.txt', 'w') as f:
    f.write('Hello, World!\n')
    f.write('This is line 2.')

# Reading from a file
with open('sample.txt', 'r') as f:
    content = f.read()
    print(content)
```
