# Statistical Analysis for Data

## Make Data-Driven Decisions
Learn statistics to extract meaningful insights from data.

```python
import pandas as pd
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
import seaborn as sns
```

## 1. Descriptive Statistics

Summarize your data.

```python
# Sample data
np.random.seed(42)
data = np.random.normal(loc=100, scale=15, size=1000)

# Calculate descriptive statistics
print('Descriptive Statistics:')
print(f'Mean: {np.mean(data):.2f}')
print(f'Median: {np.median(data):.2f}')
print(f'Mode: {stats.mode(data)[0]:.2f}')
print(f'Standard Deviation: {np.std(data):.2f}')
print(f'Variance: {np.var(data):.2f}')
print(f'Min: {np.min(data):.2f}')
print(f'Max: {np.max(data):.2f}')
print(f'Range: {np.max(data) - np.min(data):.2f}')
```

## 2. Correlation Analysis

Measure relationships between variables.

```python
# Create sample data
np.random.seed(42)
age = np.random.randint(20, 60, 100)
salary = age * 2000 + np.random.randint(-20000, 20000, 100)

# Calculate correlation
correlation = np.corrcoef(age, salary)[0, 1]
print(f'Correlation between Age and Salary: {correlation:.3f}')

# Statistical test
slope, intercept, r_value, p_value, std_err = stats.linregress(age, salary)
print(f'R-squared: {r_value**2:.3f}')
print(f'P-value: {p_value:.6f}')
print(f'Regression equation: Salary = {slope:.2f} * Age + {intercept:.2f}')
```

## 3. Hypothesis Testing

Test if differences are statistically significant.

```python
# Sample: Compare two groups
group1 = np.random.normal(loc=100, scale=15, size=100)
group2 = np.random.normal(loc=105, scale=12, size=100)

# T-test
t_statistic, p_value = stats.ttest_ind(group1, group2)
print('Independent T-Test:')
print(f'T-statistic: {t_statistic:.3f}')
print(f'P-value: {p_value:.4f}')

if p_value < 0.05:
    print('Result: Significant difference (p < 0.05)')
else:
    print('Result: No significant difference (p >= 0.05)')
```

## 4. Chi-Square Test

Test for independence between categorical variables.

```python
# Contingency table
data = np.array([[20, 30],  # Group A: [Success, Failure]
                 [25, 25]]) # Group B: [Success, Failure]

chi2, p_value, dof, expected = stats.chi2_contingency(data)
print('Chi-Square Test:')
print(f'Chi-square statistic: {chi2:.3f}')
print(f'P-value: {p_value:.4f}')
print(f'Degrees of freedom: {dof}')

if p_value < 0.05:
    print('Result: Variables are dependent (p < 0.05)')
else:
    print('Result: Variables are independent (p >= 0.05)')
```

## 5. Confidence Intervals

Estimate population parameters with confidence.

```python
# Sample data
data = np.random.normal(loc=100, scale=15, size=100)

# Calculate 95% confidence interval for mean
mean = np.mean(data)
std_err = stats.sem(data)  # Standard error of mean
confidence_level = 0.95
margin_error = std_err * stats.t.ppf((1 + confidence_level) / 2, len(data) - 1)

ci_lower = mean - margin_error
ci_upper = mean + margin_error

print(f'Mean: {mean:.2f}')
print(f'95% Confidence Interval: [{ci_lower:.2f}, {ci_upper:.2f}]')
```

## 6. Probability Distributions

Understand common distributions.

```python
# Normal distribution
x = np.linspace(-4, 4, 100)
y = stats.norm.pdf(x)

plt.figure(figsize=(10, 6))
plt.plot(x, y, linewidth=2, color='#667eea', label='Normal Distribution')
plt.fill_between(x, y, alpha=0.3)
plt.title('Normal Distribution (Mean=0, StdDev=1)', fontsize=14, fontweight='bold')
plt.xlabel('Value')
plt.ylabel('Probability Density')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
