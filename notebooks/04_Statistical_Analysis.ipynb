## Setup

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

np.random.seed(42)
sns.set_style('whitegrid')
```

## 1. Descriptive Statistics

```python
# Sample data
data = np.array([45, 50, 52, 48, 55, 53, 49, 51, 47, 54, 56, 52, 50, 48, 53])

# Calculate statistics
mean = np.mean(data)
median = np.median(data)
std_dev = np.std(data)
variance = np.var(data)
min_val = np.min(data)
max_val = np.max(data)
range_val = max_val - min_val

print(f"Mean: {mean:.2f}")
print(f"Median: {median:.2f}")
print(f"Standard Deviation: {std_dev:.2f}")
print(f"Variance: {variance:.2f}")
print(f"Min: {min_val}, Max: {max_val}")
print(f"Range: {range_val}")
```

```python
# Using pandas for quick stats
df = pd.DataFrame({
    'Score': data
})

print(df.describe())
```

## 2. Probability Distributions

```python
# Normal distribution
data_normal = np.random.normal(100, 15, 1000)

plt.figure(figsize=(12, 5))

# Histogram
plt.subplot(1, 2, 1)
plt.hist(data_normal, bins=30, density=True, alpha=0.7, color='#667eea', edgecolor='black')

# Add normal distribution curve
mu, sigma = np.mean(data_normal), np.std(data_normal)
x = np.linspace(data_normal.min(), data_normal.max(), 100)
plt.plot(x, stats.norm.pdf(x, mu, sigma), 'r-', linewidth=2, label='Normal Distribution')
plt.title('Normal Distribution')
plt.xlabel('Value')
plt.ylabel('Density')
plt.legend()

# Q-Q plot
plt.subplot(1, 2, 2)
stats.probplot(data_normal, dist="norm", plot=plt)
plt.title('Q-Q Plot')

plt.tight_layout()
plt.show()
```

## 3. Hypothesis Testing

```python
# One-sample t-test
# Test if the mean is significantly different from 50
data = np.array([45, 50, 52, 48, 55, 53, 49, 51, 47, 54, 56, 52, 50, 48, 53])

t_stat, p_value = stats.ttest_1samp(data, 50)

print("One-Sample T-Test")
print(f"Null Hypothesis: Mean = 50")
print(f"T-statistic: {t_stat:.4f}")
print(f"P-value: {p_value:.4f}")

if p_value < 0.05:
    print("Result: Reject null hypothesis (statistically significant)")
else:
    print("Result: Fail to reject null hypothesis")
```

```python
# Two-sample t-test
group1 = np.array([48, 50, 52, 51, 49, 53, 50])
group2 = np.array([55, 57, 56, 58, 54, 59, 57])

t_stat, p_value = stats.ttest_ind(group1, group2)

print("\nTwo-Sample T-Test")
print(f"Group 1 mean: {np.mean(group1):.2f}")
print(f"Group 2 mean: {np.mean(group2):.2f}")
print(f"T-statistic: {t_stat:.4f}")
print(f"P-value: {p_value:.4f}")

if p_value < 0.05:
    print("Result: Groups have significantly different means")
else:
    print("Result: No significant difference between groups")
```

## 4. Correlation Analysis

```python
# Create sample data
age = np.array([25, 30, 35, 28, 32, 40, 29, 38, 45, 33])
salary = np.array([50000, 55000, 65000, 60000, 62000, 75000, 58000, 70000, 80000, 63000])

# Pearson correlation
correlation, p_value = stats.pearsonr(age, salary)

print(f"Pearson Correlation: {correlation:.4f}")
print(f"P-value: {p_value:.4f}")
print(f"Interpretation: ")
if correlation > 0.7:
    print("  Strong positive correlation")
elif correlation > 0.3:
    print("  Moderate positive correlation")
else:
    print("  Weak correlation")
```

```python
# Visualize correlation
plt.figure(figsize=(10, 6))
plt.scatter(age, salary, s=100, alpha=0.6, color='#667eea')

# Add regression line
z = np.polyfit(age, salary, 1)
p = np.poly1d(z)
plt.plot(age, p(age), "r--", linewidth=2, label=f'r = {correlation:.3f}')

plt.title('Age vs Salary Correlation', fontsize=14, fontweight='bold')
plt.xlabel('Age')
plt.ylabel('Salary ($)')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## 5. Linear Regression

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error

# Prepare data
X = age.reshape(-1, 1)
y = salary

# Create and fit model
model = LinearRegression()
model.fit(X, y)

# Make predictions
y_pred = model.predict(X)

# Calculate metrics
r2 = r2_score(y, y_pred)
rmse = np.sqrt(mean_squared_error(y, y_pred))

print(f"Linear Regression Results")
print(f"Slope: {model.coef_[0]:.2f}")
print(f"Intercept: {model.intercept_:.2f}")
print(f"R² Score: {r2:.4f}")
print(f"RMSE: {rmse:.2f}")

# Predict salary for age 36
predicted_salary = model.predict([[36]])[0]
print(f"\nPredicted salary for age 36: ${predicted_salary:,.2f}")
```

## 6. Chi-Square Test (Categorical)

```python
# Create contingency table
from scipy.stats import chi2_contingency

# Example: Relationship between Department and Promotion
contingency_table = np.array([
    [20, 5],    # Sales: Promoted, Not Promoted
    [15, 15],   # IT: Promoted, Not Promoted
    [10, 5]     # Finance: Promoted, Not Promoted
])

chi2, p_value, dof, expected = chi2_contingency(contingency_table)

print(f"Chi-Square Statistic: {chi2:.4f}")
print(f"P-value: {p_value:.4f}")
print(f"Degrees of Freedom: {dof}")

if p_value < 0.05:
    print("Result: Variables are significantly related")
else:
    print("Result: No significant relationship between variables")
```
